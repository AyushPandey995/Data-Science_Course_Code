```python
from bs4 import BeautifulSoup
```

```python
with open ("HTML_Sample_Files/Page1.html", "r") as f:
    content = f.read()
```

```python
soup = BeautifulSoup(content, "html.parser")
```

```python
h3s = soup.find_all("h3")
```

```python
for h3 in h3s:
    print(h3.find('a')['title'])
    
```

```python
articles = soup.find_all('article')
```

```python
data = []
for article in articles:
    title = article.find('h3').find('a')['title']
    #price = article.find('p')['class'] #if we use this then we will get the first occurance of 'p' which will not work 
    price = article.select_one('p.price_color').text.split('£')[1]
    Star_rating = article.find('p')['class'][1] #This can be also done as article.select_one('p.star-rating')
    print(Star_rating)
    data.append([title, Star_rating, price+"£"])
```

```python
data
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data, columns=['Book Title', 'Star-Rating(Out of 5)', 'Price'])
```
