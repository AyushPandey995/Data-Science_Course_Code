```python
from sklearn.tree import DecisionTreeClassifier

features = [[160, 0],
            [130, 0],
            [80,1],
            [110, 1]]

labels = ['apple', 'apple', 'orange', 'orange']

clf = DecisionTreeClassifier()
clf = clf.fit(features, labels)

def prediction(a, b):
    model_prediction = clf.predict([[a, b]])
    print(f"The item is {model_prediction}") 
    
    
```

```python
prediction(40,0)
```

```python
prediction(110,1)
```

```python
prediction(140,1)
```

```python
prediction(190,1)
```
